**Nested looping with for loop**

```r
child = c("child1", "child2", "child3")
sweet = c("sweet1", "sweet2", "sweet3")

for (x in child){
  for(y in sweet){
    print(paste(x, y))
  }
}
```

**The break command**

```r
drink = c("coffee", "lemonade", "tea", "juice")
for(x in drink) {
  if(x == "tea"){
    break
  }
  print(x)
}
```

**The next command**

```r
drink = c("coffee", "lemonade", "tea", "juice")
for(x in drink){
  if(x == "lemonade"){
    next
  }
  print(x)
}
```

**The While loop**

```r
i = 1
while(i < 10){
  print(i ^ 2)
  i = i + 2
}
```

```r
sumfunction = function() {
  sum = 0
  number = as.integer(readline(prompt = "Please select any number less than 25: "))
  while(number <= 25)
  {
    sum = sum + number
    number = number + 1
  }
  print(paste("The sum of numbers received from the While loop: ", sum))
}
```

```r
sumfunction()
```

**The repeat loop**

```r
i = 1
repeat {
  print(i ^ 2)
  i = i + 2
  if(i > 10)
  break
}
```

```r
i = 1
repeat {
  i = i + 1
  if(i < 10) next
  print(i ^ 2)
  if(i >= 13) break
}
```

**Functions**

```r
#functions (single variable)
abc = function(x){
  x ^ 2
}
```

```r
abc (3)
abc(6)
abc(9)
```

```r
#functions(two variables)
abc = function(x,y){
  x ^ 2 + y ^ 2
}
```

```r
abc(3, 4)
abc(10, 10)
abc(-2, -3)
```

```r
abc = function(x,y){
  sin(x) ^ 2 + cos(x) ^ 2 + x
}
```

```r
abc(9)
abc(99)
abc(-15)
```
